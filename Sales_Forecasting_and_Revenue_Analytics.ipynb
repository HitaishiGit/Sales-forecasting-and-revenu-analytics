```python

for df in [train, test]:
    df['Year'] = df['Date'].dt.year
    df['Month'] = df['Date'].dt.month
    df['Day'] = df['Date'].dt.day
    df['WeekOfYear'] = df['Date'].dt.isocalendar().week.astype(int)
    df['IsWeekend'] = (df['DayOfWeek'] >= 6).astype(int)

print("Date features created.")
display(train[['Date', 'Year', 'Month', 'Day',
               'WeekOfYear', 'IsWeekend']].head())
```

```python
for df in [train, test]:
    df['CompetitionOpen'] = (
        (df['Year'] - df['CompetitionOpenSinceYear']) * 12
        + (df['Month'] - df['CompetitionOpenSinceMonth'])
    )

    df['CompetitionOpen'] = df['CompetitionOpen'].clip(lower=0)

    df['CompetitionOpen'] = df['CompetitionOpen'].where(
        (df['CompetitionOpenSinceYear'] > 0) &
        (df['CompetitionOpenSinceMonth'] > 0),
        0
    )

print("Competition features created.")
display(train[['CompetitionOpenSinceYear',
               'CompetitionOpenSinceMonth',
               'CompetitionOpen']].head(10))
```

```python
for df in [train, test]:
    df['PromoActive'] = df['Promo']
    df['HasPromo2'] = df['Promo2']

print("Promotion features created.")

display(train[['Promo', 'Promo2', 'PromoActive', 'HasPromo2']].head())
```

```python
categorical_cols = [
    'StoreType',
    'Assortment',
    'StateHoliday',
    'PromoInterval'
]

train = pd.get_dummies(
    train,
    columns=categorical_cols,
    drop_first=True,
    dtype=int
)

test = pd.get_dummies(
    test,
    columns=categorical_cols,
    drop_first=True,
    dtype=int
)

# Ensure both datasets have identical feature columns
train, test = train.align(test, join='left', axis=1, fill_value=0)

print("Categorical encoding completed.")
print("Training shape:", train.shape)
print("Testing shape:", test.shape)
```

```python
train = train.sort_values('Date').reset_index(drop=True)

split_date = train['Date'].quantile(0.8)

train_data = train[train['Date'] < split_date].copy()
valid_data = train[train['Date'] >= split_date].copy()

print("Split date:", split_date)
print("Training data:", train_data.shape)
print("Validation data:", valid_data.shape)
print("Training period:", train_data['Date'].min(), "to", train_data['Date'].max())
print("Validation period:", valid_data['Date'].min(), "to", valid_data['Date'].max())
```

```python
y_train = train_data['Sales']
y_valid = valid_data['Sales']

drop_cols = ['Sales', 'Customers', 'Date', 'Id',
             'PromoActive', 'HasPromo2']

X_train = train_data.drop(
    columns=drop_cols,
    errors='ignore'
)

X_valid = valid_data.drop(
    columns=drop_cols,
    errors='ignore'
)

print("X_train:", X_train.shape)
print("X_valid:", X_valid.shape)
print("y_train:", y_train.shape)
print("y_valid:", y_valid.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

model = RandomForestRegressor(
    n_estimators=100,
    max_depth=15,
    min_samples_split=10,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)

print("Random Forest training completed")
```

```python
y_pred = model.predict(X_valid)
mae = mean_absolute_error(y_valid, y_pred)
rmse = np.sqrt(mean_squared_error(y_valid, y_pred))


mask = y_valid > 0
rmspe = np.sqrt(
    np.mean(((y_valid[mask] - y_pred[mask]) / y_valid[mask]) ** 2)
) * 100

print("MAE:", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("RMSPE:", round(rmspe, 2), "%")
```

```python
from sklearn.ensemble import HistGradientBoostingRegressor

gb_model = HistGradientBoostingRegressor(
    max_iter=150,
    learning_rate=0.1,
    max_leaf_nodes=31,
    random_state=42
)

gb_model.fit(X_train, y_train)

print("Gradient Boosting training completed")
```

```python
gb_pred = gb_model.predict(X_valid)

gb_mae = mean_absolute_error(y_valid, gb_pred)
gb_rmse = np.sqrt(mean_squared_error(y_valid, gb_pred))

mask = y_valid > 0
gb_rmspe = np.sqrt(
    np.mean(((y_valid[mask] - gb_pred[mask]) / y_valid[mask]) ** 2)
) * 100

print("Gradient Boosting MAE:", round(gb_mae, 2))
print("Gradient Boosting RMSE:", round(gb_rmse, 2))
print("Gradient Boosting RMSPE:", round(gb_rmspe, 2), "%")
```
